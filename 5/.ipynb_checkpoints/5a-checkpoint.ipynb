```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import os

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
import zipfile

with zipfile.ZipFile('caltech-101.zip', 'r') as zip_ref:
    zip_ref.extractall('/content/caltech-101')
```

```python
dataset_path = "/content/caltech-101/caltech-101"
dataset = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    image_size=(224,224),
    batch_size=32
)
print("Classes: ", dataset.class_names)
```

```python
train_size = int(0.8 * len(dataset))
train_ds = dataset.take(train_size)
test_ds = dataset.skip(train_size)
print("dataset split completed")
```

```python
base_model = tf.keras.applications.VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(224, 224, 3)
)
print("VGG16 model loaded!")
```

```python
for layer in base_model.layers:
  layer.trainable=False
print("VGG16 layers frozen!")
```

```python
model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(128, activation='relu'),
    Dense(102, activation='softmax')
])
print("Custom layers added")
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Model compiled!")
```

```python
history = model.fit(
    train_ds, validation_data=test_ds,epochs=5
)
```

```python
loss, accuracy = model.evaluate(test_ds)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

```python
images, labels = next(iter(test_ds))
predictions = model.predict(images)

for i in range(5):
  plt.imshow(images[i].numpy().astype("uint8"))
  plt.title(dataset.class_names[np.argmax(predictions[i])])
  plt.show()
```

```python
for layer in base_model.layers[-4:]:
  layer.trainable=True

model.compile( optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
print("Model ready for finetuning")
```

```python
history = model.fit(
    train_ds,
    validation_data=test_ds,
    epochs=2
)
```

```python
loss, accuracy = model.evaluate(test_ds)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```
