```python
import pandas as pd
import tensorflow as tf
import re
import numpy as np
from gensim.models import Word2Vec
data = pd.read_csv(r"D:\DL\LP IV Updated Datasets\CBOW\twitter_dataset.csv")
print(data.head())
print(data.shape)
```

```python
data=data.dropna()
tweets=[]
for text in data['Text']:
    words=re.findall(r'\b[a-z]+\b',text.lower())
    tweets.append(words)
```

```python
word2Vec=Word2Vec(
    sentences=tweets,
    vector_size=100,
    min_count=2,
    sg=0,
    window=5,
    epochs=10
)
```

```python
print(word2Vec.wv['government'])
```

```python
print(word2Vec.wv.most_similar())
```
