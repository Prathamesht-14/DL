```python
import re
from gensim.models import Word2Vec
```

```python
with open('/content/document3.txt', 'r') as file:
  text = file.read().lower()
words = re.findall(r'\b[a-z]+\b', text)
print("total words: " , len(words))
print("sample words: ", words[:20])
```

```python
word2vec = Word2Vec(
    sentences = [words],
    vector_size = 100,
    window = 5,
    min_count = 1,
    sg = 0,
    epochs = 100
)
print("Vocabulary size : ", len(word2vec.wv))
```

```python
print("embedding for virus : ")
print(word2vec.wv['virus'])

print("\nSimilar words: ")
print(word2vec.wv.most_similar('virus', topn=5))
```
