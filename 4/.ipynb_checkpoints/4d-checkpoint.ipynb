```python
import pandas as pd
import numpy as np
import re

from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity

data = pd.read_csv('/content/twitter_dataset.csv')
print(data.head())
print(data.shape)
```

```python
data = data.dropna()
tweets = []

for text in data['Text']:
  words = re.findall(r'\b[a-z]+\b', text.lower())
  tweets.append(words)
print("total tweets:", len(tweets))
print("First tweet:", tweets[0])
```

```python
word2vec=Word2Vec(
    sentences = tweets,
    vector_size=100,
    window=5,
    min_count =2,
    sg=0,
    epochs = 10
)
print("vocabulary size: ", len(word2vec.wv))
```

```python
tweet_vectors = []

for words in tweets:
  vectors = []
  for word in words:
    if word in word2vec.wv:
      vectors.append(word2vec.wv[word])

  if len(vectors) == 0:
    tweet_vectors.append(np.zeros(100))
  else:
    tweet_vectors.append(np.mean(vectors,axis=0))

tweet_vectors = np.array(tweet_vectors)
print("Tweet embeddings shape: ", tweet_vectors.shape)
```

```python
query = input("Enter any topic : ").lower()
words = re.findall(r'\b[a-z]+\b', query)

vectors = []

for word in words:
  if word in word2vec.wv:
    vectors.append(word2vec.wv[word])

if vectors:
  query_vector = np.mean(vectors, axis = 0).reshape(1, -1)
  similarities = cosine_similarity(query_vector, tweet_vectors)[0]
  top = similarities.argsort()[-5:][::-1]

  for i in top:
    print(data.iloc[i]['Text'])
    print("Similarity:", round(similarities[i], 3), "\n")

else:
  print("No known words found!")
```
