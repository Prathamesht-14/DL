```python
import pandas as pd
import tensorflow as tf
import re
import numpy as np

from gensim.models import Word2Vec

data = pd.read_csv('/content/twitter_dataset.csv')
print(data.head())
print(data.shape)
```

```python
data = data.dropna()

tweets = []

for text in data['Text']:
  words = re.findall(r'\b[a-z]+\b', text.lower())
  tweets.append(words)

print("total tweets : ", len(tweets))
print('sample :', tweets[0])
```

```python
word2vec = Word2Vec(
    sentences = tweets,
    vector_size=100,
    min_count = 2,
    sg = 0,
    window = 5,
    epochs = 10
)
print("vocab size : ", len(word2vec.wv))
```

```python
print('embedding for government : ')
print(word2vec.wv['government'])
```

```python
print("\nsimilar words : ")
print(word2vec.wv.most_similar('government', topn=3))
```
